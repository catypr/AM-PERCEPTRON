```python
import matplotlib.pyplot as plt 
```

```python
pesos = [0.8, 0.3]
bias = -7.0 

dias_linha = list(range(11))

gasto_linha = [(-pesos[0] * d - bias) / pesos[1] for d in dias_linha]

hospedes = {
    "Ana": [8,7],
    "Pedro": [4,3]
}

```

```python
plt.figure(figsize=(8,6))
plt.plot(dias_linha, gasto_linha, color='blue', linestyle='--', label='Fronteira de Decisão')

# Regiões
plt.fill_between(dias_linha, gasto_linha, 30, color='green', alpha=0.1, label='Zona Upgrade(1)')
plt.fill_between(dias_linha, gasto_linha, 0, color='red', alpha=0.1, label='Zona Upgrade(0)')

# Hospedes
cores= {"Ana": "green", "Pedro": "red"}
for nome, coordenadas in hospedes.items():
    plt.scatter(coordenadas[0], coordenadas[1], color=cores[nome], s=150, edgecolors='black', zorder=5) 
    plt.text(coordenadas[0] + 0.2, coordenadas[1] - 0.5, nome, fontsize=10, fontweight='bold') 

plt.title('Fronteira de Decisão de Perceptron')
plt.xlabel('Dias Hospedados (x1)')
plt.ylabel('Gasto (X2)')
plt.xlim(0, 10)
plt.ylim(0, 30)
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```
